```python
#Ejercico de Phyton
#Clasificación de notas

def clases_nota(nota):
    if nota < 0 or nota > 5:
        return "Nota inválida"
    elif nota < 3.0:
        return "Reprueba"
    elif nota <= 4.4:
        return "Aprueba"
    else:
        return "Excelente"

# Convertimos la entrada a float para poder realizar comparaciones numéricas
nota_e = float(input("Digita la nota"))
estado = clases_nota(nota_e)
print(f"Nota: {nota_e} -> {estado}")
```

```python
# Lista de notas
notas = [2.5, 3.8, 4.7, 5.2, 3.0]
aprobados = 0

for nota in notas:
    estado = clases_nota(nota)
    print(f"Nota: {nota} -> {estado}")

    if estado in ["Aprueba", "Excelente"]:
        aprobados += 1

print("Estudiantes que aprobaron:", aprobados)
```

```python
#Carrito de compras
carrito = {
    "camiseta": (45000, 2),
    "jean": (120000, 1),
    "medias": (12000, 3)
}

subtotal = 0

for producto, datos in carrito.items():
    precio, cantidad = datos
    subtotal += precio * cantidad

print("El valor subtotal es ",subtotal)
```

```python
# Descuento del 10% al superar el valor
sub_descuento = subtotal

if subtotal > 200000:
    sub_descuento = subtotal * 0.90
    print("Subtotal con descuento",sub_descuento)

```

```python
# IVA del 19%
total = sub_descuento * 1.19

print(f"Total a pagar: ${total:,.0f}")
```

```python
#Nomina con Pandas
import pandas as pd

# DataFrame inicial_tabla
empleados = pd.DataFrame({
    "nombre": ["Pedro", "Pablo", "Luisa"],
    "area": ["Ventas", "Contabilidad", "Desarrollo"],
    "salario": [2500000, 3500000, 4500000]
})

print("=== TABLA DE EMPLEADOS ===")
print(empleados)

```

```python
# Agregar 3 empleados nuevos
nuevos = pd.DataFrame({
    "nombre": ["Pepe", "María", "Juan"],
    "area": ["Ventas", "Contabilidad", "Desarrollo"],
    "salario": [2500000, 5000000, 2800000]
})

empleados = pd.concat([empleados, nuevos], ignore_index=True)

print("=== TABLA DE EMPLEADOS ACTUALIZADA===")
print(empleados)

```

```python
# 2. Crear columna bono
empleados["bono"] = empleados.apply(
    lambda x: x["salario"] * 0.10 if x["area"] == "Ventas"
    else x["salario"] * 0.05,
    axis=1
)

# Costo total de nómina
empleados["costo_total"] = empleados["salario"] + empleados["bono"]

print("=== TABLA DE EMPLEADOS BONO + COSTO===")
print(empleados)
```

```python
# Nómina por área
nomina_area = empleados.groupby("area")["costo_total"].sum()

# Empleado con mayor salario
mayor_salario = empleados.loc[empleados["salario"].idxmax()]


print("\n=== COSTO TOTAL DE NOMINA POR AREA ===")
print(nomina_area)

```
