```python
#9
cuenta = 1200
propina = 10
amigos = 4

# 1. Regresa la cuenta más la propina

def con_propina(cuenta, porcentaje):
    propina = cuenta * (porcentaje / 100)
    total = cuenta + propina
    return total

pago_final = con_propina(cuenta, propina)
print("La cuenta total es $",pago_final)

# 2. Regresa cuánto paga cada persona

def por_persona(total, personas):
    total_per = total / personas
    return total_per

pago_per = por_persona(pago_final,amigos)
print("La cuenta por persona es $",pago_per)




# 3. Úsalas con tus datos e imprime


```

```python
#10
# 1. Pedimos el nombre (texto) y el precio (número) de cada producto
producto1 = input("Digita el primer producto: ")
precio1 = float(input(f"Digita el precio de {producto1}: "))

producto2 = input("Digita el segundo producto: ")
precio2 = float(input(f"Digita el precio de {producto2}: "))

producto3 = input("Digita el tercer producto: ")
precio3 = float(input(f"Digita el precio de {producto3}: "))

producto4 = input("Digita el cuarto producto: ")
precio4 = float(input(f"Digita el precio de {producto4}: "))

# Guardamos los datos en listas para poder recorrerlos
productos = [producto1, producto2, producto3, producto4]
precios_compra = [precio1, precio2, precio3, precio4]
# 2. Sumamos el total de los productos ingresados usando un ciclo for
total = 0
for precio in precios_compra:
    total += precio

# 3. Aplicamos descuento del 10% si el total supera los $500
descuento = 0
if total > 500:
    descuento = total * 0.10

total_pagar = total - descuento
print(f"{'PRODUCTO':<25}{'PRECIO':>15}")
# Usamos un for para recorrer e imprimir cada fila de la tabla
for i in range(len(productos)):
    print(f"{productos[i]:<25}${precios_compra[i]:>14.2f}")

print("-"*40)
print(f"{'Subtotal:':<25}${total:>14.2f}")
print(f"{'Descuento (10%):':<25}-${descuento:>14.2f}")
print("="*40)
print(f"{'TOTAL A PAGAR:':<25}${total_pagar:>14.2f}")
print("="*40)
```
